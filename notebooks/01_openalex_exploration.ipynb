```python
def reconstruct_abstract(inverted_index):
    if inverted_index is None:
        return None

    words = []

    for word, positions in inverted_index.items():
        for position in positions:
            words.append((position, word))

    words.sort()

    return " ".join(word for position, word in words)
```

```python
import requests
import pandas as pd
import time

def collect_openalex_papers(
    search_term,
    start_year=2018,
    end_year=2026,
    max_papers=1000,
    per_page=100
):
    
    url = "https://api.openalex.org/works"
    
    all_papers = []
    cursor = "*"
    
    while len(all_papers) < max_papers:
        
        params = {
            "search": search_term,
            "filter": (
                f"from_publication_date:{start_year}-01-01,"
                f"to_publication_date:{end_year}-12-31,"
                "has_abstract:true,"
                "type:article"
            ),
            "per-page": per_page,
            "cursor": cursor
        }
        
        response = requests.get(url, params=params)
        
        if response.status_code != 200:
            raise Exception(
                f"API request failed: {response.status_code}"
            )
        
        data = response.json()
        
        results = data["results"]
        
        if not results:
            break
        
        all_papers.extend(results)
        
        cursor = data["meta"]["next_cursor"]
        
        print(f"Collected {len(all_papers)} papers...")
        
        time.sleep(0.1)
    
    papers = pd.DataFrame(all_papers)
    
    return papers.head(max_papers)  
```

```python
research_areas = [
    "machine learning",
    "deep learning",
    "natural language processing",
    "computer vision",
    "large language models",
    "reinforcement learning",
    "explainable artificial intelligence"
]

all_papers = []

for area in research_areas:
    print(f"\nCollecting: {area}")
    
    area_papers = collect_openalex_papers(
        area,
        start_year=2018,
        end_year=2026,
        max_papers=1500
    )
    
    area_papers["search_area"] = area
    
    all_papers.append(area_papers)
```

```python
combined_papers = pd.concat(
    all_papers,
    ignore_index=True
)

print(combined_papers.shape)
```

```python
print("Total records:", len(combined_papers))
print("Unique papers:", combined_papers["id"].nunique())
print(
    "Duplicate records:",
    len(combined_papers) - combined_papers["id"].nunique()
)
```

```python
def reconstruct_abstract(inverted_index):
    if not inverted_index:
        return None

    words = []

    for word, positions in inverted_index.items():
        for position in positions:
            words.append((position, word))

    words.sort(key=lambda x: x[0])

    return " ".join(word for position, word in words)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(paper_embeddings)

print(similarity_matrix.shape)
```

```python
def find_similar_papers(paper_index, top_n=10):
    
    similarities = similarity_matrix[paper_index]
    
    similar_indices = np.argsort(similarities)[::-1]
    
    similar_indices = similar_indices[similar_indices != paper_index]
    
    top_indices = similar_indices[:top_n]
    
    results = clean_papers.iloc[top_indices][
        ["title", "publication_year", "cited_by_count"]
    ].copy()
    
    results["similarity"] = similarities[top_indices]
    
    return results
```
